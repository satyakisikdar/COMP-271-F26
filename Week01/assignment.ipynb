## Isomorphic Strings (LC 205)

You are given two strings, `s` and `t`, of the same length. Your task is to determine whether the two strings are **isomorphic**.

Two strings are **isomorphic** if you can replace each character in `s` with another character to obtain `t`, following these rules:

1. Each character in `s` must always map to the **same** character in `t`.
2. Different characters in `s` must map to **different** characters in `t`.
3. A character is allowed to map to itself.
4. The order of characters must remain the same.

In other words, there must be a one-to-one correspondence between the characters of `s` and the characters of `t`.

### Example

```
Input:  s = "egg", t = "add"
Output: True
```

**Explanation:**
- `'e'` maps to `'a'`
- `'g'` maps to `'d'`

Applying these mappings to `"egg"` produces `"add"`.


### Example

```
Input:  s = "foo", t = "bar"
Output: False
```

**Explanation:**
- `'f'` maps to `'b'`
- `'o'` would need to map to both `'a'` and `'r'`, which is not allowed


### Example

```
Input:  s = "paper", t = "title"
Output: True
```

**Explanation:**
Each character in `s` maps consistently to a unique character in `t`.

## Constraints

- `1 <= len(s) <= 500`
- `len(t) == len(s)`
- Both `s` and `t` consist of letter characters only.
- Your solution may **not** use a `dict` or a `set` (or any other hash-based structure) -- use plain lists, the way we practiced in class. This is what makes the problem worth doing: without a dict, "has this character been seen before" requires a linear scan you write yourself.

You may **NOT assume** that the inputs are valid strings satisfying the constraints above. Your function should return `False` if either input does not satisfy these constraints.

## What to Submit

A Python function named `are_isomorphic` with signature `are_isomorphic(s: str, t: str) -> bool` that returns `True` if the two strings are isomorphic and `False` otherwise, plus any additional helper functions you deem necessary. In addition to the [Programmer's Pact](../../programmers-pact.md) specifications, your code **cannot use** any `import` statements.

## Longest Balanced Binary Subarray (LC 525)

You are given a list of integers `nums` that contains only `0`s and `1`s. This is called a **binary array**.

Your task is to find the **maximum length** of a **contiguous subarray** that contains an **equal number of `0`s and `1`s**.

- A **subarray** is a consecutive sequence of elements from the list.
- The subarray must contain the same number of `0`s as `1`s.
- If no such subarray exists, the result should be `0`.

### Example

```
Input:  nums = [0, 1]
Output: 2
```

**Explanation:**
The entire array `[0, 1]` contains one `0` and one `1`, so its length is 2.

### Example

```
Input:  nums = [0, 1, 0]
Output: 2
```

**Explanation:**
The subarray `[0, 1]` (or `[1, 0]`) contains one `0` and one `1`. No longer contiguous subarray satisfies the condition.

### Example

```
Input:  nums = [0, 1, 1, 1, 1, 1, 0, 0, 0]
Output: 6
```

**Explanation:**
The subarray `[1, 1, 1, 0, 0, 0]` contains three `1`s and three `0`s, giving a length of 6. This is the longest such subarray.

## Constraints

- `1 <= len(nums) <= 2,000`
- Each element of `nums` is either `0` or `1`.
- The subarray must be **contiguous**; you may not reorder or skip elements.
- A straightforward (quadratic) solution -- checking every subarray -- is perfectly acceptable at this point in the course. You do not need to find the optimal linear-time approach.

You may **NOT assume** that `nums` satisfies the constraints above (e.g. it may be `None`, contain values other than `0`/`1`, or not be a list at all). Your function should return `0` if the input is invalid, exactly as it would for a valid input with no balanced subarray -- there is no separate "invalid input" signal for this problem.

## What to Submit

A Python function named `contiguous_length` with signature `contiguous_length(nums: list[int]) -> int` that returns the length of the longest contiguous subarray with an equal number of `0`s and `1`s, plus any additional helper functions you deem necessary. In addition to the [Programmer's Pact](../../programmers-pact.md) specifications, your code **cannot use** any `import` statements.

## Mirror Letters

This problem builds on the block-letter representation from [`block_letters.py`](https://github.com/satyakisikdar/COMP-271-F26/blob/main/Week01/block_letters.py) (the same module used for `mississippi.py`/`mississippi_horizontal.py` in class) -- download it and place it next to your `week01.py` before you start.

A **block letter** is a list of 10 equal-length row strings (as in `block_letters.m`, `block_letters.i`, etc.). A **word**, for this problem, is simply a list of block letters in order, e.g. `[block_letters.h, block_letters.i]` for "HI".

Your task is to compute what a word looks like held up to a mirror: the letters appear in **reverse order**, and each letter is itself **flipped left-to-right** (every row string is reversed character by character).

### Example

```
Input:  letters = [block_letters.h, block_letters.i]   # "HI"
Output: [mirrored block_letters.i, mirrored block_letters.h]
```

**Explanation:** Held up to a mirror, "HI" reads right-to-left as the mirror image of "IH": the letters swap order (I before H), and each letter's own rows are reversed left-to-right, e.g. row `"**      **"` of a letter becomes `"**      **"`, and an asymmetric row like row 0 of `block_letters.p` (`"**********"`) versus a row like row 5 (`"**        "`) becomes `"        **"`.

### Example

```
Input:  letters = []
Output: []
```

**Explanation:** An empty word mirrors to an empty word.

## Constraints

- `letters` is a list of block letters; each block letter is a list of row strings.
- All block letters in a valid input have the same number of rows, and within a single letter every row has the same width -- but this is **not guaranteed** by the caller. Your function should return `None` if this is not the case (e.g. mismatched row counts, mismatched row widths within a letter, or `letters` is not a list at all).
- You must reverse each row **manually, with a loop** -- slicing (`row[::-1]`), `reversed()`, and `.reverse()` are not allowed for this problem. The point of this problem is practicing the reversal logic itself, not Python's built-in shortcut for it. (This is checked directly on your submitted code, not just your output.)
- The only `import` allowed anywhere in `week01.py` is `import block_letters`, needed for this problem. No other imports are permitted, in this problem or the other two.

## What to Submit

A Python function named `mirror_letters` with signature `mirror_letters(letters: list[list[str]]) -> list[list[str]] | None` that returns the mirrored word as described above (a list of block letters, each a list of row strings), or `None` if the input is invalid. Plus any additional helper functions you deem necessary.
