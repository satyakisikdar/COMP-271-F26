## Problem 1: Index-Based Recursion in a Dynamic Array

The template's `DynamicArray` is a trimmed copy of Week03's, keeping only what this problem needs: a list `_underlying` that always holds `_capacity` slots, a count `_size` of how many slots hold real values, and the sentinel `_EMPTY` (`-1`) in every slot past the end.

On Day 9 the deck summed a list with `total(nums[1:])` and noted that each slice **copies** the rest of the list, and that passing an index instead avoids the copies. This problem makes you do exactly that. Neither method may slice: each recursive call passes along **positions**, and every call works on the same `_underlying`.

Implement two methods. The template already gives you `reverse()` and `count(value)`, which call yours on the whole array.

- `reverse_between(lo, hi)` reverses the stored values at positions `lo` through `hi` (both included), **in place**.
- `count_matches_from(value, start)` returns how many stored values at positions `start` through `len(self) - 1` equal `value`.

### Example

```
a = DynamicArray.from_values([10, 20, 30, 40, 50])   # capacity 8
a.reverse_between(1, 3)
print(a)             # [10, 40, 30, 20, 50]
a.reverse()
print(a)             # [50, 20, 30, 40, 10]
a._underlying        # [50, 20, 30, 40, 10, -1, -1, -1]
```

**Explanation:** only positions 1 through 3 change order. The three sentinel slots past the end are never touched: only positions `0` through `len(a) - 1` hold values.

### Example

```
same = a
a.reverse()
print(same)          # [10, 40, 30, 20, 50]
```

**Explanation:** `reverse_between` returns nothing; it changes the array itself. `same` and `a` are two names for **one** object, so `same` sees the change. For the same reason, never write `a = a.reverse()`: that would leave `a` holding `None`.

### Example

```
b = DynamicArray.from_values([7, 3, 7, 7, 1])
b.count(7)                    -> 3
b.count_matches_from(7, 1)    -> 2
b.count_matches_from(7, 5)    -> 0
b.count(-1)                   -> 0
```

**Explanation:** starting at position 1 skips the first `7`. Starting at `len(b)` leaves nothing to count. `b` has three sentinel slots holding `-1`, but they are not stored values, so `count(-1)` is `0`.

## Constraints

- `reverse_between` returns nothing, so it has **no** `return` statement.
- In `reverse_between`, an empty range (`lo > hi`) does nothing. A non-empty range that reaches outside positions `0` through `len(self) - 1` raises `IndexError`.
- In `count_matches_from`, a `start` at or past `len(self)` gives `0`, and a negative `start` raises `IndexError`.

## Problem 2: Peel One Piece per Call

Two functions, each of which removes one piece of its input per call, like `digit_sum` in the Day 9 handout and `reverse` and `is_pal` in the Day 9 deck.

- `digits_ascending(n)` returns `True` if every decimal digit of `n` is **at most** the digit to its right. Equal neighbors are fine.
- `collapse(text)` returns `text` with every run of repeated adjacent characters cut down to one.

### Example

```
digits_ascending(1223)   -> True
digits_ascending(1323)   -> False
digits_ascending(7)      -> True
digits_ascending(0)      -> True
digits_ascending(10)     -> False
```

**Explanation:** in `1223` each digit is at most the next one; the two `2`s are equal, which is allowed. In `1323` the `3` is followed by a smaller `2`. A single digit, `0` included, has no neighbor to be out of order with.

### Example

```
collapse("aabccca")      -> "abca"
collapse("mississippi")  -> "misisipi"
collapse("zzzz")         -> "z"
collapse("")             -> ""
```

**Explanation:** each run of the same character, however long, becomes one copy of it. The final `a` of `"aabccca"` is kept: it is not next to the first `a`s.

Write `is_sorted(values)`, which returns `True` if `values` is in non-decreasing order (each item at most the next), in that same divide, conquer, combine shape.

### Example

```
is_sorted([2, 5, 5, 8])          -> True
is_sorted([])                    -> True
is_sorted([7])                   -> True
is_sorted([1, 3, 2, 4])          -> False
is_sorted([1, 2, 5, 3, 4, 6])    -> False
```

**Explanation:** equal neighbors are allowed, and zero or one item is always in order.

```
is_sorted(list(range(5_000)))    -> True
```

## Problem 4: Binary Search for the First Match

The Day 9 deck's recursive `find` searched a sorted list by checking the middle and throwing away the half that could not hold the target. It assumed every value appeared once. Here the list may hold **repeats**, and you must return the position of the **first** one.

Write `first_index_between(nums, target, lo, hi)`, which returns the smallest index `i` with `lo <= i <= hi` and `nums[i] == target`, or `-1` if there is none. The template gives you `find_first(nums, target)`, which calls yours on the whole list.

### Example

```
nums = [2, 5, 5, 5, 8, 12, 12, 23]
find_first(nums, 5)     -> 1
find_first(nums, 12)    -> 5
find_first(nums, 2)     -> 0
find_first(nums, 7)     -> -1
find_first([], 5)       -> -1
```

**Explanation:** `5` sits at positions 1, 2 and 3; the answer is the first of them.

### Example

```
first_index_between(nums, 5, 2, 7)    -> 2
first_index_between(nums, 5, 4, 7)    -> -1
find_first([4] * 100_000, 4)          -> 0
find_first(list(range(100_000)), 99_999)   -> 99999
```

**Explanation:** only positions `lo` through `hi` count, so a `5` before position `2` does not. The last two lines are there to make sure your search really halves its range: halving needs about 17 nested calls on 100,000 values, while stepping one position at a time needs up to 100,000 and runs past Python's limit of about 1000. Both lines are in the template's test block, so you will see this on your own machine.

### Trace it before you run it

Once your function works, trace your own version on paper: list the `lo` and `hi` of every call `find_first(nums, 5)` makes on the list above, and what each call returns.

## Constraints

- `first_index_between` must be recursive, as described at the top of this notebook, and must **halve** the range `lo` to `hi` on each call. Tests use lists of up to **100,000** values.
- `nums` is sorted in non-decreasing order. Callers always pass `0 <= lo` and `hi < len(nums)`, or an empty range with `lo > hi`.
- No slicing: pass positions, as `find` did. `first_index_between` must not change `nums`.
