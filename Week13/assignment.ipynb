## How Big O is graded

A stopwatch is a poor judge of Big O: a slow computer makes fast code look slow, and a fast one hides a Θ(n²) loop on small inputs. So the autograder **counts steps** instead. While your function runs, it counts:

- every line of **your** code that runs (a loop body that runs 1,000 times counts 1,000 times), and
- every comparison (`==`, `<`, `>`) made on a value the test passes in (a station name, a rider name or a number from the test's list), including the ones hidden inside Python: `x in some_list`, `.count()`, `.index()` and `sorted()` each compare against the items they walk past.

It runs your function on an input of size `n` and again on a bigger one (4 times bigger in most tests), and compares the two counts. For an input 4 times bigger, Θ(n) code takes about **4 times** as many steps; Θ(n²) code takes about **16 times** as many. The test passes when the count grows by at most 1.5 times what Θ(n) gives (6 times for an input 4 times bigger), and its feedback shows both counts. A nested-loop `has_pair` gets:

```
has_pair on n even numbers, looking for the odd total 3
  300 rides: 90,303 steps (90,303 lines of your code, 0 comparisons)
  1,200 rides: 1,441,203 steps (1,441,203 lines of your code, 0 comparisons)
  16.0 times as many steps for 4.0 times as many rides; Theta(n) grows about 4.0 times, and this test allows up to 6.0.
```

The comparisons are 0 there because `minutes[i] + minutes[j] == target` compares a new number, the sum, not one of the test's own values; every one of those 1,441,203 lines still counts.

Each step-counting test is visible on every submission, and worth points of its own inside the problem it belongs to. **It also checks the answer on the input it measures**: a function that returns the wrong thing earns nothing for being fast, since doing nothing at all takes very few steps. A 5-second time limit still stops a call that never ends.

## Problem 1: The Rider Directory

The bike share keeps a directory from each rider's username to a number, such as how many rides they have taken. The directory is a hash table with **separate chaining**, like the lecture's `ChainedHashTable` (Week 12-1), with two changes: its keys are **strings**, so it needs a string hash, and it **grows** when its chains get long.

Implement three pieces:

- `poly_hash(word, slot_count)` is the Hash Tables I deck's `poly_hash` (Week 12-1), with the number of slots passed in. Start a total at 0; for each character, multiply the total by `BASE` (31) and add the character's code, `ord(char)`; return the total's remainder when divided by `slot_count`. Because each character is weighted by its position, anagrams such as `"stop"` and `"pots"` usually land in different slots, unlike `sum_hash`.
- `ChainedTable.put(key, value)` stores `value` under `key` in the chain at slot `poly_hash(key, number of slots)`. If the key is already in that chain, replace its pair in place, keeping its position. Otherwise append the new pair to the end of the chain and add 1 to `_size`. Then, if the load factor α has **passed** `MAX_CHAIN_LOAD` (1.0), call the given `self._grow()` once. Chaining tables grow near α = 1, about one key per chain (Week 12-2). The deck's way of finding the key (a list of the chain's keys, then `in` and `.index` on it) is fine here: it searches one short chain, not the whole table.
- `ChainedTable.get(key)` returns the value stored under `key`, or `None` if it is not stored. It looks in key's own chain and nowhere else.

Each chain is a list of `(key, value)` tuples. A tuple cannot be changed, so replacing a value means writing a new tuple into the chain at the same index.

### Example

```python
poly_hash("ab", 8)           -> 1
poly_hash("ba", 8)           -> 7
poly_hash("", 11)            -> 0
poly_hash("stop", 11)        -> 6
poly_hash("pots", 11)        -> 3
poly_hash("tops", 11)        -> 1
```

**Explanation:** for `"ab"` the total is 97 × 31 + 98 = 3105, and 3105 % 8 is 1. For `"ba"` it is 98 × 31 + 97 = 3135, and 3135 % 8 is 7: the same letters, different slots. These are the Coat Check handout's numbers, and the three anagrams are from the poly_hash slide; both are Week 12-1. An empty word never changes the total, so it lands in slot 0.

### Example

```python
riders = ChainedTable()                 # 11 empty chains
for name in ["ana", "ben", "cy", "dee"]:
    riders.put(name, len(name))
riders.put("ben", 9)                    # ben is already stored: replace
riders.get("ben"), riders.get("cy"), riders.get("zoe"), riders.size()
    -> 9 2 None 4
# put 7 more new names, each with value 1:
riders.size(), riders.slot_count()      -> 11 11
riders.put("lu", 1)
riders.size(), riders.slot_count(), riders.get("ana")
    -> 12 23 3
```

**Explanation:** replacing `ben`'s value adds no key, so the size stays 4. With 11 keys in 11 slots, α = 1.0 has not **passed** 1.0, so the table does not grow. The 12th key makes α = 12 / 11 ≈ 1.09, so `put` calls `_grow`, which moves every pair into `next_prime(22)` = 23 chains. `get("ana")` still finds 3 afterward, because `_grow` hashed every key again with 23 slots.

## Problem 2: The Bike Dock

The dock records which building each bike is parked at, in a hash table with **open addressing and linear probing**: each slot holds at most one bike. `_keys[i]` is the bike number in slot `i` (`None` means empty) and `_values[i]` its dock, the two parallel lists of the Hash Tables II deck's `ProbingHashTable` (Week 12-2). A bike's home slot is `key % number of slots`.

The template gives `_find_slot(key)`, the deck's walk: it starts at the home slot and steps right (wrapping from the last slot to 0) until it reaches the slot holding `key` or an empty slot. It also gives `get`, and `checks(key)`, which counts the slots a lookup looks at: the home slot, each slot holding another key, and the last one, whether it holds `key` or is the empty slot that ends a miss. Implement three methods:

- `put(key, value)` finds the key's slot with `self._find_slot(key)`. If that slot is empty, the key is new: add 1 to `_size`. Write the key and value into the slot. Then, if α has **passed** `MAX_PROBE_LOAD` (0.7), call `self._resize()` once.
- `_resize()` is **rehashing**, as on the Moving Day handout (Week 13-1): replace both lists with `next_prime(GROWTH_FACTOR * old number of slots)` empty slots, then visit the **old** slots in order 0, 1, 2, ... and place each stored key, with its value, at the slot `self._find_slot` now gives it. Old slot numbers mean nothing in the new table: `k % 17` is not `k % 7`.
- `longest_run()` returns the length of the table's **longest run**: the most full slots in a row. A run may **wrap** from the last slot around to slot 0, as in the Moving Day handout, where slots 14, 15, 16, 0 and 1 form one run of 5. An empty table returns 0, and a table with every slot full returns the number of slots. Long runs are what make linear probing slow (Traffic Jams, Week 12-2): a miss that lands in a run of length L checks L + 1 slots. Watch the wrap: a run that crosses the end must not be counted as two short ones. One way in is to start just after an empty slot, so no run is cut where you begin.

### Example

The bikes and sizes are the Moving Day handout's, so you can check your work against its solutions.

```python
dock = ProbingTable()                   # 7 empty slots
for bike in [14, 42, 31, 15]:
    dock.put(bike, "Cudahy")
dock.slots()                            -> [14, 42, 15, 31, None, None, None]
dock.longest_run(), ProbingTable().longest_run()
    -> 4 0
```

**Explanation:** 14 and 42 both have home 0, so 42 steps to slot 1. 31 lands at home 3. 15 has home 1, which is full, so it steps to 2. Slots 0 to 3 are full: one run of 4. An empty table has no run at all.

### Example

```python
dock.put(50, "Damen")                   # home 1, walks to slot 4: 5 bikes in 7 slots
dock.slot_count(), dock.size(), dock.longest_run()
    -> 17 5 4
dock.slots()
    -> [50, None, None, None, None, None, None, None, 42, None, None, None, None, None, 14, 15, 31]
dock.put(33, "Halas")
dock.put(42, "Piper")                   # 42 is already stored: new dock, same slot
dock.get(33), dock.get(42), dock.get(99), dock.size()
    -> Halas Piper None 6
dock.longest_run(), dock.checks(48)     -> 5 6
```

**Explanation:** after `put(50)`, α = 5 / 7 ≈ 0.71 has passed 0.7, so the table resizes to `next_prime(14)` = 17 slots. Rehashing in old slot order: 14 goes home to 14, 42 to 8, 15 to 15, then 31 (home 14) walks to 16, and 50 (home 16) wraps around to slot 0. The full slots are now 14, 15, 16 and 0, one run of 4 that wraps around the end. 33 has home 16 and walks 16, 0, 1, so the run grows to 5: slots 14, 15, 16, 0 and 1. 48 (home 14) lands in that run and is a miss: the given `checks` counts its 5 full slots plus the empty slot 2, six in all, which is L + 1 for a run of L = 5.

## Constraints

- `put` and `_resize` find slots with the given `self._find_slot`. `_resize` must **not** call `put`: `put` would count every key a second time and could resize in the middle of a resize.
- `_resize` leaves `_size` unchanged and visits the old slots in order 0, 1, 2, ...; the order matters whenever two keys collide in the new table.

- No `dict` or `set`.
- Tests use up to **2,000** bikes, with bike numbers up to 1,000,000, and count the steps a run of `n` puts takes: it must grow like Θ(n).

Implement two functions:

- `insertion_sort(names)` sorts a list of names into alphabetical order, **in place**, with insertion sort: for `i` from 1 up, take `names[i]`, shift each bigger name before it one slot right, and drop it into the gap. Python compares strings alphabetically with `<` and `>` (capital letters before lowercase). It returns nothing.
- `rank_stations(rides)` returns every station in `rides` once, the busiest first; stations with the same number of rides come out in alphabetical order. Do it in three steps:
    1. **Count** each station's rides in a `ChainedTable` from Problem 1 (so finish Problem 1 first), and keep a list of the stations in the order each first appears.
    2. **Scatter**: bucket `c` holds the stations with exactly `c` rides. No station has more rides than `len(rides)`, so `len(rides) + 1` buckets are enough.
    3. **Gather** from the biggest count down to 1, sorting each bucket with your `insertion_sort` before adding it to the result.

### Example

```python
names = ["Piper", "Damen", "Halas", "Cudahy"]
insertion_sort(names)
names                                   -> ['Cudahy', 'Damen', 'Halas', 'Piper']
rides = ["Halas", "Damen", "Piper", "Damen", "Cudahy", "Halas",
         "Damen", "Mundelein", "Piper"]
rank_stations(rides)                    -> ['Damen', 'Halas', 'Piper', 'Cudahy', 'Mundelein']
rank_stations([]), rank_stations(["Halas"])
    -> [] ['Halas']
```

## Problem 4: Make It Linear

A day pass covers exactly two rides if their minutes add up to the pass's limit. The template gives `has_pair_slow(minutes, target)`, which answers whether **two different rides** add up to `target` by trying every ride with every later one:

```python
for i in range(len(minutes)):
    for j in range(i + 1, len(minutes)):
        if minutes[i] + minutes[j] == target:
            found = True
```

The inner loop shrinks, but as the Big O deck (Week 11-2) showed, comparing each item with every later one still costs (n − 1) + (n − 2) + ... + 1 = n(n − 1)/2 steps: Θ(n²). For 200,000 rides that is about 20 billion additions.

Write `has_pair(minutes, target)`, which returns the same answer in **Θ(n) time on average**. Here is all the `set` syntax you need:

```python
seen: set[int] = set()       # an empty set (not {}, which is an empty dict)
seen.add(12)                 # add a value; adding it again changes nothing
12 in seen                   # True, in Θ(1) on average
25 in seen                   # False
```

 Walk the list **once**, keeping the minutes seen so far in a Python `set`. A ride pairs with an earlier one exactly when `target` minus its minutes is already in the set. `in` and `add` on a set are Θ(1) on average, because a `set` is a hash table (Week 12-2). Check a ride against the set **before** adding it, so a ride never pairs with itself.

### Example

```python
minutes = [12, 30, 7, 25, 18]
has_pair_slow(minutes, 37), has_pair(minutes, 37), has_pair(minutes, 50)
    -> True True False
has_pair([25], 50), has_pair([25, 25], 50), has_pair([], 0)
    -> False True False
evens = list(range(0, 400_000, 2))      # 200,000 rides
has_pair(evens, 3), has_pair(evens, 399_996)
    -> False True
```

**Explanation:** 30 + 7 = 37, so the first list has a pair for 37 (12 + 25 is another; walking left to right finds 30 + 7 first), but no two of its rides make 50: 25 would need a second 25. A single ride cannot pair with itself, so `[25]` has no pair for 50, while `[25, 25]` does. Every number in `evens` is even, so no two add up to the odd 3, and finding that out means walking the whole list. Your version answers both calls in a fraction of a second; the slow version takes several minutes on the first.
